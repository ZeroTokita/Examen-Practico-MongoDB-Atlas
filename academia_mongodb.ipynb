```python

from pymongo import MongoClient
from urllib.parse import quote_plus
from getpass import getpass
from pprint import pprint

usuario = input("Usuario de MongoDB: ").strip()
contrasena = getpass("Contrasena de MongoDB: ")
host = input("Host del cluster: ").strip()

uri = (
    f"mongodb+srv://{quote_plus(usuario)}:"
    f"{quote_plus(contrasena)}@{host}/"
    "?retryWrites=true&w=majority"
)

cliente = MongoClient(
    uri,
    serverSelectionTimeoutMS=15000
)

cliente.admin.command("ping")

db = cliente["academia"]
coleccion = db["estudiantes"]


print("Conexion exitosa con MongoDB Atlas")
print("Base de datos:", db.name)
print("Coleccion:", coleccion.name)


```

```python

# CONSULTA 1 - TODOS LOS ESTUDIANTES

from pprint import pprint

for estudiante in coleccion.find():
    pprint(estudiante)
```

```python

# CONSULTA 2 - PROYECCION

for estudiante in coleccion.find(
    {},
    {
        "_id": 0,
        "nombre": 1,
        "correo": 1
    }
):
    pprint(estudiante)
```

```python

# CONSULTA 3 - MAYORES DE 20 ANOS

for estudiante in coleccion.find(
    {"edad": {"$gt": 20}},
    {"_id": 0, "nombre": 1, "edad": 1}
):
    pprint(estudiante)
```

```python

# CONSULTA 4 - FILTRO EN ARREGLOS

for estudiante in coleccion.find(
    {"cursos.nombre": "Bases de Datos"},
    {"_id": 0, "nombre": 1}
):
    pprint(estudiante)
```

```python

# CONSULTA 5 - CALIFICACIONES

for estudiante in coleccion.find(
    {
        "cursos.calificacion": {
            "$gte": 80
        }
    },
    {"_id": 0, "nombre": 1, "cursos": 1}
):
    pprint(estudiante)
```

```python

# CONSULTA 6 - AGREGAR CURSO

resultado = coleccion.update_one(
    {"correo": "luis@mail.com"},
    {
        "$push": {
            "cursos": {
                "nombre": "MongoDB",
                "duracion_horas": 25,
                "calificacion": 0,
                "progreso": 0
            }
        }
    }
)

print("Documentos modificados:", resultado.modified_count)

pprint(
    coleccion.find_one(
        {"correo": "luis@mail.com"},
        {"_id": 0}
    )
)
```

```python

# CONSULTA 7 - ACTUALIZAR CALIFICACION

resultado = coleccion.update_one(
    {
        "correo": "ana@mail.com",
        "cursos.nombre": "Bases de Datos"
    },
    {
        "$set": {
            "cursos.$.calificacion": 95
        }
    }
)

print("Documentos modificados:", resultado.modified_count)

pprint(
    coleccion.find_one(
        {"correo": "ana@mail.com"},
        {"_id": 0, "nombre": 1, "cursos": 1}
    )
)
```

```python

# CONSULTA 8 - ELIMINAR ESTUDIANTE

resultado = coleccion.delete_one(
    {"correo": "temporal@mail.com"}
)

print("Documentos eliminados:", resultado.deleted_count)
print("Estudiantes restantes:", coleccion.count_documents({}))
```

```python

# CONSULTA 9 - ESTUDIANTES CON VARIOS CURSOS

total = coleccion.count_documents(
    {"cursos.1": {"$exists": True}}
)

print("Estudiantes con mas de un curso:", total)
```

```python

# CONSULTA 10 - OPERADOR IN

for estudiante in coleccion.find(
    {
        "cursos.nombre": {
            "$in": ["SQL", "MongoDB"]
        }
    },
    {"_id": 0, "nombre": 1}
):
    pprint(estudiante)
```

```python

# EXTRA - OPERADORES AND, GT Y LT

for estudiante in coleccion.find(
    {
        "$and": [
            {"edad": {"$gt": 20}},
            {"edad": {"$lt": 24}}
        ]
    },
    {"_id": 0, "nombre": 1, "edad": 1}
):
    pprint(estudiante)
```
